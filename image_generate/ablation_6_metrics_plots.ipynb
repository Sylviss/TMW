```python
def plot_six_metrics_overlay(
    df_mask1, df_mask2, x_col, title, label1="mask_type=1", label2="mask_type=2",
    figsize=(12, 4.5), show_errorbar=True
 ):
    fig, axes = plt.subplots(1, 2, figsize=figsize)
    axes = axes.flatten()

    metric_display_map = {"MMD": "KRD", "FD": "FRD"}
    x_label_map = {"lambda_ts": r"$\lambda_D$", "eps_threshold": r"$\lambda^*$"}
    x_label = x_label_map.get(x_col, x_col)
    label_fontsize = 14

    for i, metric in enumerate(metrics):
        ax = axes[i]

        # Use categorical x-axis labels for clearer comparison.
        x1 = df_mask1[x_col].astype(str)
        x2 = df_mask2[x_col].astype(str)

        if show_errorbar:
            ax.errorbar(
                x1,
                df_mask1[metric],
                yerr=df_mask1[f"{metric}_std"],
                fmt="o-",
                capsize=4,
                linewidth=2,
                markersize=5,
                label=label1,
            )
            ax.errorbar(
                x2,
                df_mask2[metric],
                yerr=df_mask2[f"{metric}_std"],
                fmt="s--",
                capsize=4,
                linewidth=2,
                markersize=5,
                label=label2,
            )
        else:
            ax.plot(
                x1,
                df_mask1[metric],
                "o-",
                linewidth=2,
                markersize=5,
                label=label1,
            )
            ax.plot(
                x2,
                df_mask2[metric],
                "s--",
                linewidth=2,
                markersize=5,
                label=label2,
            )
        
        metric_display = metric_display_map.get(metric, metric)
        ax.set_xlabel(x_label, fontsize=label_fontsize)
        ax.set_ylabel(metric_display, fontsize=label_fontsize)
        ax.tick_params(axis="x", rotation=30)
        ax.grid(True, alpha=0.3)
        ax.legend()

    fig.tight_layout()
    return fig

# Version 1: with error bars
fig1 = plot_six_metrics_overlay(
    df_lambda_m1,
    df_lambda_m2,
    x_col="lambda_ts",
    title="Ablation Sweep: lambda_ts (mask_type 1 vs 2, eps_threshold=0.2) [Error Bar]",
    show_errorbar=True,
 )

fig2 = plot_six_metrics_overlay(
    df_eps_m1,
    df_eps_m2,
    x_col="eps_threshold",
    title="Ablation Sweep: eps_threshold (mask_type 1 vs 2, lambda_ts=0.5) [Error Bar]",
    show_errorbar=True,
 )

# Version 2: mean only (no std/error bars)
fig3 = plot_six_metrics_overlay(
    df_lambda_m1,
    df_lambda_m2,
    x_col="lambda_ts",
    title="Ablation Sweep: lambda_ts (mask_type 1 vs 2, eps_threshold=0.2) [Mean Only]",
    show_errorbar=False,
 )

fig4 = plot_six_metrics_overlay(
    df_eps_m1,
    df_eps_m2,
    x_col="eps_threshold",
    title="Ablation Sweep: eps_threshold (mask_type 1 vs 2, lambda_ts=0.5) [Mean Only]",
    show_errorbar=False,
 )

plt.show()
```
